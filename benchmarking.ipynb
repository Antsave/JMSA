### 1. Sphere Benchmark (f1) 

```python
# sphere benchmark, f1, unimodal
def sphere(x): 
    return np.sum(x**2)
```

### 2. Absolute Sum + Absolute Product (f2)

```python
# absolute sum + absolute product, f2, unimodal
def asap(x):
    x = np.array(x)
    return np.sum(np.abs(x)) + np.prod(np.abs(x))
```

### 3. Rastrigin Function (f9)

```python
# rastrigin function, f9, multimodal w many local minima
def rastrigin(x): 
    return np.sum(x**2 - 10 * np.cos(2 * np.pi * x) )
```

### 4. Griewank Function (f11)

```python
# griewank function, f11, multimodal w many local minima
def griewank(x): 
    x = np.array(x)
    sum_term = np.sum(x**2) / 4000
    prod_term = np.prod(np.cos(x / np.sqrt(np.arange(1, len(x) + 1))))
    return sum_term - prod_term + 1
```

### 5. Six Hump Camel (f16)

```python
# six hump camel, f16, multimodal w few local minima
def f16(x): 
    x1 = x[0]
    x2 = x[1]
    return (
        4 * x1**2
        - 2.1 * x1**4
        + (x1**6) / 3
        + x1 * x2
        - 4 * x2**2
        + 4 * x2**4
    )
```

### 6. Branin Function (f17)

```python
# branin function, f17, multimodal w few local minima
def f17(x):
    x1, x2 = x[0], x[1]
    pi = np.pi

    term1 = x2 - (5.1 / (4 * pi**2)) * x1**2 + (5 / pi) * x1 - 6
    term2 = 10 * (1 - 1 / (8 * pi)) * np.cos(x1)

    return term1**2 + term2 + 10
```
